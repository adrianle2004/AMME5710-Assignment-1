## Image Histograms and Histogram Equalisation
During the week 1 lecture, we discussed explored image histograms and algorithms for contrast stretching and histogram equalisation. The following code snippets demonstrate the use of related functions in OpenCV.

```python
# Image Histograms
im1_gray = cv2.imread('example_images_week1/apples.jpg', cv2.IMREAD_GRAYSCALE) # read-in and convert to grayscale

# Calculate image histogram values
hist = cv2.calcHist([im1_gray],[0],None,[256],[0,256]) # ([img],[chanels],mask,bin_count,range)

# Plot the histogram next to the image
plt.subplot(121)
plt.imshow(im1_gray, 'gray', vmin=0, vmax=255)

plt.subplot(122)
plt.plot(hist)
plt.xlim([0,256])

plt.tight_layout()
```

```python
# Histogram Equalisation
im_pout = cv2.imread('example_images_week1/pout.tif', cv2.IMREAD_GRAYSCALE)

# contrast adjust the image
A = 75.0 # some stretching parameters
B = 170.0
C = 0.0
D = 255.0 
im_adjusted = ((D-C)/(B-A))*(im_pout-A)+C # apply contrast adjustment
im_adjusted = np.clip(im_adjusted, 0, 255).astype('uint8') # clip to [0,255] and convert to uint8

# histogram equalise the image
im_equalised = cv2.equalizeHist(im_pout)

# Plotting
plt.subplot(231)
plt.imshow(im_pout,'gray',vmin=0,vmax=255)

plt.subplot(232)
plt.imshow(im_adjusted,'gray',vmin=0,vmax=255)

plt.subplot(233)
plt.imshow(im_equalised,'gray',vmin=0,vmax=255)

plt.subplot(234)
hist_orig = cv2.calcHist([im_pout],[0],None,[256],[0,256])
plt.plot(hist_orig)
plt.xlim([0,256])
plt.xlabel('Original')

plt.subplot(235)
hist_adjusted = cv2.calcHist([im_adjusted],[0],None,[256],[0,256])
plt.plot(hist_adjusted)
plt.xlim([0,256])
plt.xlabel('Contrast Adjusted')

plt.subplot(236)
hist_equalised = cv2.calcHist([im_equalised],[0],None,[256],[0,256])
plt.plot(hist_equalised)
plt.xlim([0,256])
plt.xlabel('Histogram Equalised')

plt.tight_layout()
```

```python
# Spatially Adaptive Histogram Equalisation
# using CLAHE (Contrast Limited Adaptive Histogram Equalisation)

im_mountain = cv2.imread('example_images_week1/mountain.png', cv2.IMREAD_GRAYSCALE)

# CLAHE
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))
im_equalised2 = clahe.apply(im_mountain)

# Plotting
plt.subplot(121)
plt.imshow(im_mountain,'gray',vmin=0,vmax=255)
plt.xlabel('Original')

plt.subplot(122)
plt.imshow(im_equalised2,'gray',vmin=0,vmax=255)
plt.xlabel('CLAHE Adjusted')

plt.tight_layout()
```

## Image Thresholding
Thresholding is the process of assigning pixels in an image to discrete classes based on whether the pixel’s value is above or below a given threshold. During the Week 1 lecture we explored methods for thresholding including Otsu's method. The following code snippets demonstrate functions used for thresholding in OpenCV.

```python
# Thresholding (using standard numpy array functions)
im_thr = np.logical_and( (im1[:,:,2] > 180), (im1[:,:,1] < 180) ) # using standard numpy

im_temp = im1.copy()
im_temp[~im_thr, :] = (0,0,0) # mask-out (set to black) non-True pixels

plt.subplot(121)
plt.imshow(im_thr.astype(int),'gray')

plt.subplot(122)
plt.imshow(cv2.cvtColor(im_temp, cv2.COLOR_BGR2RGB))
```

```python
def chess_position(frame: cv2.Mat, row: int, col: int):
    # This function takes in a row and column index (0-7) and returns the corresponding chess position in algebraic notation.
    # For example, row=0, col=0 corresponds to 'a8', row=7, col=7 corresponds to 'h1'.
    
    # Validate input
    if row < 0 or row > 7 or col < 0 or col > 7:
        raise ValueError("Row and column indices must be between 0 and 7.")
    
    # Each square is 90x90 pixels, so we can calculate the pixel coordinates based on the row and column indices.
    x = col * 90
    y = row * 90
    
    detected_chess = frame.copy() # Copy the original frame to draw on
    
    chess_square = frame[y:y + 90, x:x + 90] # Extract the square from the image
    
    cv2.rectangle(detected_chess, (x, y), (x + 90, y + 90), (255, 0, 0), 5) # Draw a rectangle on the image
    
    chess_hist = cv2.calcHist([chess_square], [0], None, [256], [0, 256]) # Calculate the histogram of the square
    
    # If the square have a white piece, the histogram will have a peak at high intensity values (close to 255). If the square is empty or has a black piece, the histogram will have a peak at low intensity values (close to 0).   
    if (chess_hist[255] > 750): # This threshold value can be adjusted based on the specific images being used
        print("White piece detected at position: row = ", row, ", col = ", col)
    elif (chess_hist[0] > 750): # This threshold value can be adjusted based on the specific images being used
        print("Black piece detected at position: row = ", row, ", col = ", col)
    else:
        print("Empty square detected at position: row = ", row, ", col = ", col)

    return detected_chess, chess_square, chess_hist

# Write your code here, or in a separate python script
img_chess = cv2.imread('example_images_week1/chess.png')
img_chess = cv2.cvtColor(img_chess, cv2.COLOR_BGR2RGB) # convert to RGB for displaying with matplotlib

img_detect = img_chess.copy()

row = 0 # row index (0-7)
col = 0 # column index (0-7)

img_detect, chess_square, chess_hist = chess_position(img_detect, row, col) # Call the function to draw the rectangle on the image
plt.imshow(img_detect)
plt.plot()
```

## Exercise 2: Histogram Equalisation
Open up the example images 'fish001.jpg', 'fish002.png' and 'underwater001.png'. Images taken underwater are typically dark, exhibit poor contrast and have a blue-green colour cast (due to attenuation of red light in water).

```python
def contrast_adjustment(image, A, B, C, D):
    """
    Adjust the contrast of an image using linear contrast stretching.
    
    Parameters:
    - image: Input image (numpy array).
    - A: Lower bound of the input intensity range.
    - B: Upper bound of the input intensity range.
    - C: Lower bound of the output intensity range.
    - D: Upper bound of the output intensity range.
    
    Returns:
    - im_adjusted: Contrast adjusted image (numpy array).
    """
    # Apply contrast adjustment formula
    im_adjusted = ((D - C) / (B - A)) * (image - A) + C
    
    # Clip values to [0, 255] and convert to uint8
    im_adjusted = np.clip(im_adjusted, 0, 255).astype('uint8')
    
    return im_adjusted

# Write your code here, or in a separate python script
img_fish1 = cv2.imread('example_images_week1/fish001.jpg', cv2.IMREAD_GRAYSCALE)
img_fish2 = cv2.imread('example_images_week1/fish002.png', cv2.IMREAD_GRAYSCALE)
img_water1 = cv2.imread('example_images_week1/underwater001.png', cv2.IMREAD_GRAYSCALE)

# contrast adjust the image
A = 60.0 # some stretching parameters
B = 170.0
C = 0.0
D = 255.0 

# Equalised 
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8,8))

fish1_adjusted = contrast_adjustment(img_fish1, A, B, C, D) # apply contrast adjustment
fish1_equalised = cv2.equalizeHist(img_fish1) # histogram equalisation
fish1_adaptive_equalised = clahe.apply(img_fish1) # adaptive histogram equalisation

fish2_adjusted = contrast_adjustment(img_fish2, A, B, C, D) # apply contrast adjustment
fish2_equalised = cv2.equalizeHist(img_fish2) # histogram equalisation
fish2_adaptive_equalised = clahe.apply(img_fish2) # adaptive histogram equalisation

water1_adjusted = contrast_adjustment(img_water1, A, B, C, D) # apply contrast adjustment
water1_equalised = cv2.equalizeHist(img_water1) # histogram equalisation
water1_adaptive_equalised = clahe.apply(img_water1) # adaptive histogram equalisation

plt.subplot(331)
plt.imshow(cv2.cvtColor(fish1_adjusted, cv2.COLOR_BGR2RGB))

plt.subplot(332)
plt.imshow(cv2.cvtColor(fish2_adjusted, cv2.COLOR_BGR2RGB))

plt.subplot(333)
plt.imshow(cv2.cvtColor(water1_adjusted, cv2.COLOR_BGR2RGB))

plt.subplot(334)
plt.imshow(cv2.cvtColor(fish1_equalised, cv2.COLOR_BGR2RGB))

plt.subplot(335)
plt.imshow(cv2.cvtColor(fish2_equalised, cv2.COLOR_BGR2RGB))

plt.subplot(336)
plt.imshow(cv2.cvtColor(water1_equalised, cv2.COLOR_BGR2RGB))

plt.subplot(337)
plt.imshow(cv2.cvtColor(fish1_adaptive_equalised, cv2.COLOR_BGR2RGB))

plt.subplot(338)
plt.imshow(cv2.cvtColor(fish2_adaptive_equalised, cv2.COLOR_BGR2RGB))

plt.subplot(339)
plt.imshow(cv2.cvtColor(water1_adaptive_equalised, cv2.COLOR_BGR2RGB))
```

## Exercise 3: Thresholding
Load the colour image 'apples.jpg' and develop code to segment the image into three classes: background, red apple, green apple using thresholding operations. 
Plot the segmented image side-by-side with the original. For the segmented image, produce a graph were red apple pixels are coloured red, green apples pixels green and background pixels white.


```python
def calc_histogram(image):
    """
    Plot the histogram of an image.
    
    Parameters:
    - image: Input image (numpy array).
    - title: Title for the histogram plot.
    """
    hist_r = cv2.calcHist([image], [2], None, [256], [0, 256])
    hist_g = cv2.calcHist([image], [1], None, [256], [0, 256])
    hist_b = cv2.calcHist([image], [0], None, [256], [0, 256])
    
    return hist_r, hist_g, hist_b

# Write your code here, or in a separate python script
img_apples = cv2.imread('example_images_week1/apples.jpg')

# Thresholding (using standard numpy array functions)
red_apple = np.logical_and( (img_apples[:,:,2] > 180), (img_apples[:,:,1] < 180) ) # using standard numpy
green_apple = np.logical_and( (img_apples[:,:,1] > 180), (img_apples[:,:,2] < 220) ) # using standard numpy

im_red = img_apples.copy()
im_red[~red_apple, :] = (0,0,0) # mask-out (set to black) non-True pixels
segment_red = im_red[400:500, 200:300]
hist_red_r, hist_red_g, hist_red_b = calc_histogram(segment_red)
cv2.rectangle(im_red,(200,400),(300,500),(255,255,0),3) # show mask location


im_green = img_apples.copy()
im_green[~green_apple, :] = (0,0,0) # mask-out (set to black) non-True pixels
segment_green = im_green[200:300, 200:300]
hist_green_r, hist_green_g, hist_green_b = calc_histogram(segment_green)
cv2.rectangle(im_green,(200,200),(300,300),(255,255,0),3) # show mask location


im_background = img_apples.copy()
im_background[(red_apple | green_apple), :] = (0,0,0) # mask-out (set to black) non-True pixels
segment_background = im_background[0:90, 0:90]
hist_background_r, hist_background_g, hist_background_b = calc_histogram(segment_background)
cv2.rectangle(im_background,(0,0),(90,90),(255,255,0),3) # show mask location


plt.subplot(241)
plt.imshow(cv2.cvtColor(img_apples, cv2.COLOR_BGR2RGB))
plt.title('Original Apples')
plt.axis('off')

plt.subplot(242)
plt.imshow(cv2.cvtColor(im_red, cv2.COLOR_BGR2RGB))
plt.title('Red Apples')
plt.axis('off')

plt.subplot(243)
plt.imshow(cv2.cvtColor(im_green, cv2.COLOR_BGR2RGB))
plt.title('Green Apples')
plt.axis('off')

plt.subplot(244)
plt.imshow(cv2.cvtColor(im_background, cv2.COLOR_BGR2RGB))
plt.title('Background')
plt.axis('off')

plt.subplot(246)
plt.plot(hist_red_g, color='green')
plt.plot(hist_red_b, color='blue')
plt.plot(hist_red_r, color='red')
plt.title('Red apple Histogram')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')

plt.subplot(247)
plt.plot(hist_green_b, color='blue')
plt.plot(hist_green_r, color='red')
plt.plot(hist_green_g, color='green')
plt.title('Green apple Histogram')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')

plt.subplot(248)
plt.plot(hist_background_r, color='red')
plt.plot(hist_background_g, color='green')
plt.plot(hist_background_b, color='blue')
plt.title('Background Histogram')
plt.xlabel('Pixel Value')
plt.ylabel('Frequency')

plt.show()
```
